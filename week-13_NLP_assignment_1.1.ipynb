```python
import joblib
import pandas as pd

# ==============================================================================
# 1. LOAD THE TRAINED PIPELINE MODEL
# ==============================================================================
model_path = "best_whatsapp_classifier.pkl"

try:
    model = joblib.load(model_path)
    print("Model 'whatsapp_nb_model.pkl' loaded successfully!\n")
except FileNotFoundError:
    print(f"Error: Could not find '{model_path}'. Make sure the file is in the same directory.")
    exit()

# ==============================================================================
# 2. PREDICT ON SINGLE / CUSTOM MESSAGES
# ==============================================================================
sample_messages = [
    "Where are you guys?",
    "missed voice call",
    "Haha that is so funny!",
    "Call me when you reach home",
    "ok fine"
]

print("=== 1. PREDICTIONS FOR SAMPLE MESSAGES ===")
# Generate predictions
predictions = model.predict(sample_messages)

# Generate probability scores for each class (sender)
probabilities = model.predict_proba(sample_messages)
classes = model.classes_

for msg, pred, probs in zip(sample_messages, predictions, probabilities):
    # Find confidence score for predicted sender
    confidence = max(probs) * 100
    print(f"Message   : '{msg}'")
    print(f"Predicted : {pred.upper()} (Confidence: {confidence:.2f}%)")
    print("-" * 50)


# ==============================================================================
# 3. PREDICT ON AN ENTIRE NEW CHAT FILE
# ==============================================================================
print("\n=== 2. APPLYING MODEL TO NEW CHAT DATAFRAME ===")

# Example: Assuming 'df' is loaded from wf.PreProcessing("NewChat.txt")
# (Using a mock dataframe here for demonstration)
new_chat_df = pd.DataFrame({
    'Date': ['12/05/2024', '12/05/2024'],
    'Time': ['14:20', '14:22'],
    'Text': ['Are we meeting today?', 'Yes see you at 5']
})

# Run batch prediction on 'Text' column
new_chat_df["Predicted_Sender"] = model.predict(
    new_chat_df["Text"].fillna("").astype(str)
)

print(new_chat_df[["Date", "Time", "Text", "Predicted_Sender"]])
```
